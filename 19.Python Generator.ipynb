## 19. Python Generator

✅ What is Generator in Python??

 * A `Generator` is a special type of function that produce one value at a time, instad of producing all values at a time.

 * A generator does `not stores all result in memory` like a list.

 * This is called:

   * ✅ Lazy Evaluation
   * ✅ Memory Efficient Iteration
### Normal Function VS Generator Function
 * A Normal Function returns a value using return.

 * Once it returns, the funciton stop permanently.

```python
# Normal Function
def normal():
    return 10
normal()
```

⭐ Generator Function
* A generator used yield instead of return 
* It pauses execution and returns later if required.

```python
def gen():
    yield 10
print(gen())
```

The generator function returns a generator object, not the value.
* To get the value:

```python
g = gen()
next(g)
```

###  What does yield do?
The yield keyword:
* ✅ Produces a value
* ✅ Pause the function states
* ✅ Saves local variable
* ✅ Resumes from the same point when called again.

    

#### ✅ Example: Generator with Multiple Yield




```python
def my_generator():
    yield 1
    yield 2
    yield 3
    yield 4

g = my_generator()
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

### Generator Execution Flow()

```python
def test():
    print("Start")
    yield 100
    print("Middle")
    yield 200
    print("End")
    yield 300
test() # i.e generator object
```

```python
g = test()
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

### 📌 Explanation:
 * First next() runs untill first yield
 * Second next() after first yield
 * Third next() resumes after second yield
 
So, Generator Works like pause-resume machine.

### Generator with Loop

```python
def numbers(n):
    for i in range(1,n+1):
        yield i

g = numbers(5)
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

```python
def numbers(n):
    for i in range(1,n+1):
        yield i
g = numbers(5)

for value in g:
    print(value)
```

✅ Why Generators are Memory Efficient?

⭐  List Example(Consumes Memory)

```python
def list_numbers(n):
    return [i for i in range(n)]
lst = list_numbers(1000000)
print(len(lst))
```

Generator Example 

```python
def gen_numbers(n):
    for i in range(n):
        yield i
g = gen_numbers(1000000)
print(g)
print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

This does NOT stores all numbers. If produces them one by one.

List Expression

```python
lst = [x*x for x in range(5)]
print(lst)
```

Using this function in Generator then

```python
gen = (x*x for x in range(5))
print(gen)
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

### Key Difference: List VS Generator Expression
----

|Feature|List|Generator|
|----|----|----|
|Memory|High|Very Low|
|Speed|Faster Access|Slightly slower|
|Storage|Stores all values|Stores only one at a time.|
|Use Case| Small data set|Large data set|

---
### yield VS return
---
|Keyword|Behavior|
|---|---|
|return|Ends function permantly|
|yield|Pauses function and resumes later|

Generator can accept input dynamically using send()

```python
def myGen():
    x = yield "Hello"  # put value in variable x
    yield x  # display value
g = myGen()
print(next(g)) # start Generator
print(g.send(50)) # send value
```

Notes:

* First next() start generator
* send(50): assign value 50 in x variable

## Generator with throw()

You can raise exception inside generator

```python
def gen():
    try:
        yield 1
        yield 2
    except ValueError:
        yield "ValueError Occurred."
g = gen()
print(next(g))
print(next(g))
print(g.throw(ValueError))
```
