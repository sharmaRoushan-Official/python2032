## 8. Loops in Python

### 1. What is a for Loop?

 * A `for loop` is used to iterate(traverse) over a sequence and execute a set of statements repeatedly. 

```python
# 1,10

for i in range(1,10):
    print(i)
```

```python
range(10)
```

```python
range(1,10)
```

### 2. Syntax of for Loop



### Explanation: 
 * for -> Keyword
 * variable: takes each value one by one.
 * in -> keyword to access elements from sequence
 * sequence -> list, tuple, string, range, dict, set
 * : -> start of loop body
 * Indentation -> defines loop block

### 3. Flow of Execution

```python
for x in [10,20,30]:
    print(x)
```

1. x = 10 -> print(x)
2. x = 20 -> print(x)
3. x = 30 -> print(x)
4. loop ends

### Using enumerate() in for Loop

* Used when you need index+value

```python
names = ["Raj","Amit","Rohit","Ankit"]

for index,value in enumerate(names):
    print(index,value)
```

```python
names = ["Raj","Amit","Rohit","Ankit"]

for a,b in enumerate(names):
    print(a,b)
```

```python
names = ["Raj","Amit","Rohit","Ankit"]

for x,y in enumerate(names):
    print(x,y)
```

### break and continue in for loop

```python
for i in range(1,11):  # first condition will be cheked then display
    if i == 6:
        break
    print(i)
```

```python
for i in range(1,11): # first display value then condition checked.
    print(i)
    if i == 6:
        break
```

### 18. Continue statement in for Loop

```python
for i in range(1,11):
    if i == 5:
        continue
    print(i)  # if i == 5 then print skip
```

```python
for i in range(1,11):
    print(i)  # if i == 5 then print skip
    if i == 5:
        continue
    
```

### While Loop 

* A while `loop` in Python is used to repeat a block of code as long as a condition is True.

### ✅ 1. Definition
* The while loop checks a condition.
* If the condition is True, the loop runs
* It keeps running unitl the conditon becomes False

### ✅ 2. Syntax


### ✅ 3. Flow of Execution(Working)

📌 steps:
1. Condition is checked first.
2. If condition is True -> code inside loop executes.
3. After execution, condition is checked again.
4. Loop continues until condition becomes False.
5. When condition is False -> Loop stops.

```python
i = 1  # start
j = 5  # stop

while i<j:  # 1<=5 , 4<5
    print(i) # 1, 2, 3, 4
    i = i+1  # increment value 
```

```python
1<5: True
2<5: True
3<5: True
4<5: True
5<5 : False
```
