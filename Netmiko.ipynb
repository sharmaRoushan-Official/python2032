## Netmiko

Netmiko is one of the most important Python Libraries used in Network Automation.

It is mainly used for automating tasks on routers, switches, and firewalls using SSH[Secure Shell].

---
1. What is Netmiko?
   * Netmiko is a Python library that provides a simplified SSH Connection to network devices.
  
It is buit on the top of `Paramiko`

📌 In simple word

`Paramiko` is a general SSH library.
`Netmiko` is a network-device automation SSH library. 

## 2. Why Netmiko was Developed?

Paramiko can connect to any SSH device, but network device have such issue:

* Problems with Paramiko for Network Devices

* ❌ Network decice required Prompt detection
* ❌ Need to handle ---more-- pagination
* ❌ Need to handle enable mode
* ❌ Need to handle config mode
* ❌ Different vendors have different CLI behavior

So, Netmiko was developed to solve all these issue.

## Netmiko Features 

Netmiko provide built-in suppoert 

* ✅ SSH connection to network devices
* ✅ Automatic prompt detection
* ✅ Automatic paging disable(terminal lenght 0)
* ✅ Enable mode support
* ✅ Config mode support
* ✅ Multi-vendor support
* ✅ Command execution with clean output
* ✅ Timeout and error handling
* ✅ Easy configrations sending
* ✅ Sessing logging

## 1. What is Paramiko

Paramiko is a Python library that implements the SSHv2 Protocol.

📌 Simple Meaning

Paramiko allows Python program to login to remote machines using SSH and run commands.

## Why Paramiko is Used

Before Paramiko, remote login and automation required manual SSH:

But in automation, we want to Python to do this automatically.

So paramiko helps in :
* ✅ Server automation
* ✅ network automation

What is SSH

* remote login

### Paramiko Components 

Paramiko is made of multiple internal components:

A). SSHClient: Used to create SSH Connection

B). Transport: Handles the SSH Session, encryption, key exchange

C). Channel: used to send/receive data(like terminal session)

D). SFTPClient: Used for Secure File Transfer Protocols Client.

```python
import paramiko

host = "192.168.0.101"
username = "user"
password = "123"

# Create SSH client instance
client = paramiko.SSHClient()

# Automatically add the host key (use with caution in production)
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

# Connect to the remote host
client.connect(hostname=host, username=username, password=password)

# Execute a command on the remote host
stdin, stdout, stderr = client.exec_command("ls")

# Read and decode the output
print(stdout.read().decode())

# Always close the connection when done
client.close()
```

## 6. Executing Commands Remotely

✅ Run Remote Linux Commands



```python
ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh.connect(hostname="192.168.0.254",username="cisco",password="cisco@123")

# Execute command remotely
# stdin,stdout,stderr = ssh.exec_command("ls -l") this command used in linux
stdin,stdout,stderr = ssh.exec_command("show version")

# Read Command Output
output = stdout.read().decode()

# Read error if any
error = stderr.read().decode()

print("Output:\n",output)
print("Error:\n",error)
ssh.close()
```

## Explaination

exec_command() # it return 3 object

|stdout|Command output|

```python
# Running Multiple Commands

commands = [
    "show version",
    # "show ip interface brief",
    # "show running-config",
    # "show ip route"
    
]

ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())
ssh.connect(hostname="192.168.0.254",username="cisco",password="cisco@123")

for cmd in commands:
    print(f"\nRunning:{cmd}")
    stdin,stdout,stderr = ssh.exec_command(cmd)
    print(stdout.read().decode())
    print(stderr.read().decode())
```

```python
import paramiko
import time

host = "192.168.0.254"
username = "cisco"
passwrod = "cisco@123"

ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh.connect(hostname=host, username=username, password=password)

shell = ssh.invoke_shell()
time.sleep(1)

commands = [
    "terminal length 0",
    "show ip interface brief",
    "show version"
]

for cmd in commands:
    shell.send(cmd + "\n")
    time.sleep(2)  # wait for output

    output = shell.recv(65535).decode()
    print(f"Command: {cmd}")
    print(output)

ssh.close()
```

### Capturing Errors(stderr)

```python
import paramiko
import time

host = "192.168.0.254"
username = "cisco"
passwrod = "cisco@123"

ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh.connect(hostname=host, username=username, password=password)

stdin,stdout,stderr = ssh.exec_command("show version")

error = stderr.read().decode()
# print(error)
if error:
    print("Error:",error)
else:
    print(stdout.read().decode())
```

## Checking Exit status



```python
import paramiko
import time

host = "192.168.0.254"
username = "cisco"
passwrod = "cisco@123"

ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh.connect(hostname=host, username=username, password=password)

stdin,stdout,stderr = ssh.exec_command("show clock")
print(stdout.read().decode())

exit_status = stdout.channel.recv_exit_status()
print("Status",exit_status)
if exit_status == 0:
    print("Command executed Successfully")
else:
    print("Command Failed")
```

```python
import paramiko
import time

ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh.connect("192.168.0.254", username="cisco", password="cisco@123")

channel = ssh.invoke_shell()

commands = [
    "enable",
    "terminal length 0",
    "show ip interface brief",
    "show version",
    "show vlan brief"
]

for cmd in commands:
    channel.send(cmd + "\n")
    time.sleep(2)

output = channel.recv(65535).decode()
print(output)

ssh.close()
```
