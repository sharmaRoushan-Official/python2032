## 16. Lambda in Python

Q. What is Lambda?

A `lambda function` is a small anonymous(unnamed) function.


syntax:

lambda arguments:expression

* No def keyword
* single expression only
* Return value automatically

## 2. What is map()?

map() applies a function to each item of an iterable(like list, tuple)

```python
# user defined function 
def sumTwo(x,y):
    return x+y
sumTwo(10,12)
```

```python
# anonymous function: function without name
(lambda x,y:x+y)(10,12)
```

```python
sumTwo(10,45)
```

```python
(lambda x,y:x+y)(10,45)
```
