## 4. Decorator Concept 

A decorator is basically:
* A function that takes another function
* Adds extra code
* Returns the modified function

### 5. Decorator Without @ Syntax

```python
### 5. Decorator Without @ Syntax

def decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper
```

```python
def hello():
    print("Hello world")
```

```python
hello = decorator(hello)
hello()
```

### 6. Decorator with @ Syntax (Most Common)

```python
def decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper
```

```python
@decorator
def hello():
    print("Hello world")
hello()
```

### 8.  Decorator with *args and **kwargs (More professional)



```python
def decorator(func):
    def wrapper(*args, **kwargs): # parameter 
        print("Before function call")
        result = func(*args,**kwargs)  # argument
        print("After function call")
        return result
    return wrapper
```

```python
@decorator
def add(a,b):
    return a+b
add(10,20)
```

```python
def decorator(func):
    def wrapper(*args, **kwargs): # parameter 
        print("Function is starting")
        value = func(*args,**kwargs)  # argument
        print("Function is finished")
        return value
    return wrapper
```

```python
@decorator
def multiplication(a,b):
    return a*b
print(multiplication(5,4))
```

```python
import time
def decorator(func):
    def wrapper(*args, **kwargs): # parameter 
        start = time.time()
        result = func(*args,**kwargs)  # argument
        end = time.time()
        print(f"Execution Time: {end-start:.5f}")
        return result
    return wrapper
```

```python
@decorator
def big_task():
    time.sleep(3)
    print("Task Done")
big_task()
```

```python
## 11. Real use case 2: Logging Decorator


def logger(rohit):
    def wrapper(*args,**kwargs):
        print(f"Calling Function:{rohit.__name__}") # this function show the name of accutal function name
        return rohit(*args, **kwargs)
    return wrapper
```

```python
@logger
def login(user):
    print(f"{user} logged in")
login("Ram")
```

```python
@logger
def loginUser(user):
    print(f"{user} logged in")
loginUser("Siddharth")
```

### Real Use Case 3: Authentication Decorator(Django Style)



```python
def login_required(func):
    def wrapper(user):
        if user == "admin" or user == "guest":
            return func(user)
        else:
            return "Access Denied"
    return wrapper
```

```python
@login_required
def dashboard(user):
    return f"Welcome to dashboard {user}"

print(dashboard("admin"))
```

```python
print(dashboard("guest"))
```

```python
print(dashboard("Ankit"))
```

## Multiple Decorators


```python
# this function only convert string value into upper case
def upper_case(func):
    def wrapper():
        return func().upper()
    return wrapper

# this function add exclamation mark on the function
def exclamation(func):
    def wrapper():
        return func() + "!!!!"
    return wrapper

```

```python
@exclamation
@upper_case
def message():
    return "hello"
message()
```

### Class Decorator

```python
from functools import wraps
```

```python
help(wraps)
```

```python
def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper
```

```python
@decorator
def hello():
    """This is the function"""
    print("Hello")
print(hello.__name__) # prints hello
print(hello.__module__)
print(hello.__qualname__)
print(hello.__doc__)
print(hello.__annotations__)
print(hello.__type_params__)
```

```python
hello()
```

```python
def greet(name: str, age: int = 25) -> str:
    return f"Hello {name}, you are {age} years old."

# Accessing the annotations directly
print(greet.__annotations__)
# Output: {'name': <class 'str'>, 'age': <class 'int'>, 'return': <class 'str'>}
```

```python
# class decorator

def class_decorator(cls):
    cls.company = "TCS"
    return cls
```

```python
@class_decorator
class Employee:
    def __init__(self,name):
        self.name = name

e = Employee("Ravi")
print(e.company)
```
