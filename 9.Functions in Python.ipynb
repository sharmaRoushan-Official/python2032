## 9. Functions in Python

#### 1. What is Function?
 * A `function` is a reusable block of code that performs a specific task.

#### Why we use functions?
 * Avoid code repetation
 * Make code readable
 * Improve modularity
 * Easier debugging
 * Reusability

### 2. Creating a function 

✅ Syntax:

def function_name():
    
    # block of code 

```python
def greet():  # define function with function name
    print("Hello, Welcome to Python!!")  # block of code 
```

```python
greet()  # function calling
```

## 3 functions with parameters(Arguments)

✅ Parameters:
Value passed into a function


✅ Syntax:

def function_name(parameter):

    # block of code
    
function_name(argument) # function calling

```python
def greet(name):  # name-parameter
    print("Hello",name)

greet("Rohit")  # Rohit - argument
```

```python
greet("Ankit")
```

```python
greet("Shyam")
```

```python
greet("Ram")
```

```python
def sum1(x):  # x is parameter
    a1 = x+10
    b1 = x-20
    c1 = x*10
    d1 = x/10
    print(f"The result will be {a1},{b1},{c1},{d1}: if you can pass x value {x}")

sum1(10)
```

```python
sum1(20)
```

```python
sum1(-50)
```

```python
sum1(80)
```

### 4. Function with Return Value

✅ return statement:

Returns value back to caller:

```python
def add(a,b):
    return a+b

result = add(5,10)
print(result)
```

```python
x = add(10,-10)
print(x)
```

### 5. Types of Functions Arguments

1️⃣ Positional Arguments


```python
def student(name,age):
    print(name,age)
student("Shyam",25)
```

2️⃣ Keyword Arguements


```python
student(age=25,name="Rohit")
```

3️⃣ Default Arguments

```python
def greet(name="Guest"):
    print("Hello",name)
greet()
```

```python
greet("rohit")
```

```python
greet()
```

```python
greet("Ankit")
```

4️⃣ Varible Lenght Arguments

✅ *args (Non-Keyword variable arguments)

```python
def total(*number):
    print(sum(number))
total(10)
```

```python
total(10,20)
```

```python
total(10,20,-70)
```

```python
total(10,20,30,50,60,70,-50,-80)
```

✅ **kwargs(Keyword variable argument)

```python
def details(**info):
    print(info)
details(name="Ravi",age=25)
```

```python
details(name="rohit",mob="123456789",age=25)
```
