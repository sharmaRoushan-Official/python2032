## 9. Functions in Python

#### 1. What is Function?
 * A `function` is a reusable block of code that performs a specific task.

#### Why we use functions?
 * Avoid code repetation
 * Make code readable
 * Improve modularity
 * Easier debugging
 * Reusability

### 2. Creating a function 

✅ Syntax:

def function_name():
    
    # block of code 

```python
def greet():  # define function with function name
    print("Hello, Welcome to Python!!")  # block of code 
```

```python
greet()  # function calling
```

## 3 functions with parameters(Arguments)

✅ Parameters:
Value passed into a function


✅ Syntax:

def function_name(parameter):

    # block of code
    
function_name(argument) # function calling

```python
def greet(name):  # name-parameter
    print("Hello",name)

greet("Rohit")  # Rohit - argument
```

```python
greet("Ankit")
```

```python
greet("Shyam")
```

```python
greet("Ram")
```

```python
def sum1(x):  # x is parameter
    a1 = x+10
    b1 = x-20
    c1 = x*10
    d1 = x/10
    print(f"The result will be {a1},{b1},{c1},{d1}: if you can pass x value {x}")

sum1(10)
```

```python
sum1(20)
```

```python
sum1(-50)
```

```python
sum1(80)
```

### 7. Scope of Variable

✅ Local Variable 

```python
def test():
    x = 10   # x is a variable it is local variable 
    print(x)
    print(x)
test()
```

```python
print(x)  # x is not defined 
```

✅ Global Variable 

```python
x = 100 # x is global variable

def test():
    print(x) # Accessing inside the function
test()
```

```python
print(x) # Accessing outside the function 
```

### 8. Recursion

✅ Function Calling itself.

```python
# 8 = 1,2,3,4,5,6,7

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)  # 3 * 2

factorial(4)
```

### 9. Nested Functions

✅Function inside function

```python
def outer():
    def inner():
        print("Inside inner function")
    inner()
outer()
```

```python
def outer():
    def inner():
        print("Inside inner function")
        def inner2():
            print("Inner2 Function calling")
            def inner3():
                print("Inner3 function calling")
            inner3()
        inner2()
    inner()
outer()
```

✅ 10. Function as Object

In Python, functions are first-class objects

✅ Assign Function to variable

```python
def greet():
    print("Hello")
x = greet()
x
```

✅ Pass function as argument


```python
def display(func):
    func()  # greet()

display(greet)
```

```python
def sum2():
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    sum1 = num1+num2
    print(f"The sum of two numbers {num1} and {num2} is {sum1}")
sum2()
```

```python
def callingFunc(func):
    func() # sum2()
callingFunc(sum2)
```
