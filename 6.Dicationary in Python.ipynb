# 6. Dictionary in Python

 * A __dictionary__ is a collection of key-value pairs.
   * __Each key is unique.__
   * __Each key maps to a value.__
   * __Dictionaries are mutable(changeable).__
   * __They are written using curly braces {}__

```python
student = {1}
print(student,type(student))
```

```python
student = {
    "name":"Ankit",
    "age":21,
    "course":"Python"
}
print(student, type(student))
```

```python
student = {"name":"Ankit","age":21,"course":"Python"}
print(student, type(student))
```

* name,age,course -> keys
* Ankit,21,Python --> Values

### 16. Real-World Examples

Example 1 - Student record

```python
student = {
    'id':101,
    "name":"Ankit",
    "marks":85
}
print(student)
```

Example 2: Shopping Card

```python
cart = {
    "laptop":50000,
    "mouse":500,
    "keyboard":1500,
    "Toffee":52
}
print(cart)

# we can use DOM method [Document object Module]
x = cart['laptop']
print(x)
y = cart['mouse']
print(y)

x1 = cart['keyboard']

y1 = cart['Toffee']

sum1 = x+y+x1+y1
print(sum1)
```

```python
sum1 = 0
for values in cart.values():
    sum1 = sum1 + values
    print(values)
print("Total values:",sum1)
```
