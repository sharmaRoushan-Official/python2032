# 6. Dictionary in Python

 * A __dictionary__ is a collection of key-value pairs.
   * __Each key is unique.__
   * __Each key maps to a value.__
   * __Dictionaries are mutable(changeable).__
   * __They are written using curly braces {}__

```python
student = {1}
print(student,type(student))
```

```python
student = {
    "name":"Ankit",
    "age":21,
    "course":"Python"
}
print(student, type(student))
```

```python
student = {"name":"Ankit","age":21,"course":"Python"}
print(student, type(student))
```

* name,age,course -> keys
* Ankit,21,Python --> Values

### 2. Why We Use Dictionaries

Use dictionary when data has a `relationship between key and value.`

Example Uses:

|Key|Value|
|---|---|
|name|Ankit|
|age|21|
|course|Python|


```python
employee={
    "id":101,
    "name":"Ankit",
    "salary":50000
}
print(employee,type(employee))
```

```python
employee.keys()
```

```python
employee.values()
```

```python
employee.items()
```

### 3.Characteristics of Dictionary 

Important points

1️⃣ keys must be unique

```python
data ={
    "name":"Ankit",
    "name":"Rohit"
}
'''Last value overwrites the previous one.'''
data
```

2️⃣ Keys must be immutable


__Allowed:__
`string`
`number`
`tuple`

__Not allowed:__
`list`
`dictionary`

```python
data = {
    1:"Python",
    "name":"Ankit",
    (1,2):"Tuple Key"
}
data
```

3️⃣ Value can be anything

Value can be:
 * `string`
 * `number`
 * `list`
 * `tuple`
 * `dictionary`
 * `object`

```python
data = {
    "name":"Ankit",
    "makrs":[80,90,85],
    "address":{
        "city":"Vasco da gama",
        "state":"Goa"
    }
}

print(data, type(data))
```
