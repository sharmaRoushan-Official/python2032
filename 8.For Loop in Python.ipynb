## 8. Loops in Python

### 1. What is a for Loop?

 * A `for loop` is used to iterate(traverse) over a sequence and execute a set of statements repeatedly. 

```python
# 1,10

for i in range(1,10):
    print(i)
```

```python
range(10)
```

```python
range(1,10)
```

### 2. Syntax of for Loop



### Explanation: 
 * for -> Keyword
 * variable: takes each value one by one.
 * in -> keyword to access elements from sequence
 * sequence -> list, tuple, string, range, dict, set
 * : -> start of loop body
 * Indentation -> defines loop block

### 3. Flow of Execution

```python
for x in [10,20,30]:
    print(x)
```

1. x = 10 -> print(x)
2. x = 20 -> print(x)
3. x = 30 -> print(x)
4. loop ends

### 4. for loop with range()
* The most common loop in python

```python
for i in range(5):
    print(i)
```

 range(5) generates numbers from 0 to 4


### 5. range(start,stop)

```python
for i in range(2,6):
    print(i)
```

* starts at 2
* stops before 6

### 6. range(start, stop, step)

```python
for i in range(1,10,2):  # n-1 = 2-1 = 1
    print(i)
```

* step = 2 (jumpy by 2)

### 7. Reverse value [Decending Order]

```python
for i in range(10,0,-1):
    print(i)
```

```python
for i in range(10,-1,-1):
    print(i)
```

### 8. for loop with String

```python
name = "Ankit"

for ch in name:
    print(ch)
```

### 9. for loop with List

```python
fruits = ['apple',"Banana","Mango"]

for f in fruits:
    print(f)
```

```python
for rohit in fruits:
    print(rohit)
```

### 10. for loop with tuple


```python
t = (10,20,30,40,50,60)

for item in t:
    print(item)
```

### 11. for loop with set


```python
s = {1,2,3,2,2,2,2,3,5,5,6,6,7,8,9,6}

for item in s:
    print(item)

# Set order is not fixed.
```

### 12. for loop with Dictionary

printing keys

```python
d = {'a':10,'b':20,"c":30}

for key in d:
    print(key)
```

```python
for key in d.keys():
    print(key)
```

```python
for value in d.values():
    print(value)
```

```python
for item in d.items():
    print(item)
```

```python
# Write a program to find even number between 1-20

for i in range(1,21):
    if i%2 == 0:
        print(i)
```

```python
for i in range(1,21):
    if i%2!=0:
        print(i)
```

```python
## table 

# 19x1=19
# 19x2=38
# 19x3=57

num = 19

for i in range(1,11):
    print(num,"X",i,"=",num*i)
```

```python
num = 19

for i in range(1,11):
    print(f"{num}X{i}={num*i}")
```

```python
num = 19

for i in range(1,11):
    print("%dX%d=%d"%(num,i,num*i))
```
