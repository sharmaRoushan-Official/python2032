## 1. Encapsulation

Encapsulation means wrapping data and methods together and restricting direct access to data.

* `How do you protect data inside a class?`

### Types of Access Modifiers in Python

|Modifier|Syntax|meaning|
|---|---|---|
|Public|var|Accessible everywhere|
|Protected|_var|Accessible in child class|
|Private|__var|Not directly accessible|

### Exqample - Encapsultion

1. Public Members
* Accessible everywhere
* No underscore prefix

```python
class Student:
    def __init__(self,name):
        self.name = name # public variable

    def display(self): # public method
        return f"Student Name:{self.name}"

s = Student("Rohit")
print(s.name)  # Accessible public variable
print(s.display()) # Accessible public method
```

```python
print(s.name)
```

```python
print(s.display())
```

### 2. Protect Members(_)
* Should be accessed only inside class and child class
* Defined using single underscore (_var)

⚠️ still accessible outside, but considered not recommended.

```python
class Employee:
    def __init__(self,name,age,salary):
        self.name = name  # public variable
        self.age = age   # public variable
        self._salary = salary # protected variabled

    def DetailsEmployee(self):
        print(f"Customer name:",self.name)
        print(f"Customer age:",self.age)
        print(f"Customer Salary:",self._salary)


c = Employee("Rohit",99,"12345678")
# print(c.name) # it is public
# print(c.age) # it is public
# print(c.salary) # it is prtected 
c.DetailsEmployee()
```

### 3. Private members(__)

* Accessible only inside the class
* Defined using double underscore ( __var)
* Python applies name mangling(internally changes name)


```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance # private variable

    def showBalance(self):
        return f"Balance:{self.__balance}"

b = BankAccount(1000)
#b.__balance # private variable
b.showBalance()
```

### 2. Abstraction(Hiding Implementation)

* Abstractoin means `showing only essential features` and hinding internal implementation.

* "What to show and what to hide?"

## Achieved Using:
* Abstract Classes
* Interfaces (via ABC Module)

```python
from abc import ABC, abstractmethod
```

```python
class vehicle(ABC):
    
    @abstractmethod # decorator
    def start(self):
        pass
    @abstractmethod
    def steering(self):
        pass
```

```python
class car(vehicle):
    def start(self):
        print("Car Started")

    def steering(self):
        print("Car Steering")

c =car()
c.start()
```

```python
c.steering()
```

# 1. Single Inheritance

In `Single Inheritance`, a child class inherits from only one parent class.

### 📌Structure
* Parent -> Child

### Why it is used
* Code reusbility
* Child class can use parent class properties and methods
* Easy to understand and maintain

# Real-life Example
* Parent- `Vehicle`
* Child - `Car`

```python
# parent class
class Vehicle:
    def start(self):
        print("Vehicle Started")
v = Vehicle()
v.start()
```

```python
# class child
class Car(Vehicle):
    def drive(self):
        print("Car is driving")

c = Car()
c.start()
c.drive()
```

## 2. Multiple Inheritance

In `Multiple Inheritance`, a child class inherits from more than one parent class.

📌 Structure

* Parent1     Parent2
*       \   /
*       child 

## why it is used
* When one class needs features from multiple classes
* Helps combine functionalities.

## Realife Example
* Parent1:`Camera`
* Parent2:`Phone`
* Child: `Smart Phone`

```python
class Camera: # parent 1
    def click_photo(self):
        print("Photo Clicked")


class Phone: # parent 2
    def call(self):
        print('Calling')
c = Camera()
c.click_photo()
```

```python
p = Phone()
p.call()
```

```python
# child class 

class SmartPhone(Phone,Camera):
    # pass
    def game(self):
        print("Play Video Game")
    

s = SmartPhone()
s.click_photo() # it is a feature of parent class
s.call() # it is a feature of parent class
s.game() # it is a feature of child class
```

### 3. Multilevel Inheritance

In `Multilevel Inheritance`,inheritance happsn in levels.

📌 Structure

Grandparent ➡️Parent ➡️ Child 

## Why it is used

* To represent heirarchy in real world
* Step-by-step inheritance of features

## Real-life Eample
* Grandparent: `Animal`
* Parent: `Mammal`
* Child: `Dog`
  
Dog is a Mammal, Mammal is an Animal

```python
class Animal: # Grandparent
    def eat(self):
        print("Eating.........")
class Mammal(Animal):  # parent
    def walk(self):
        print("Walking.......")
class Dog(Mammal):  # child
    def speak(self):
        print("Dog Barks......")

d = Dog()
d.eat()
d.walk()
d.speak()
```

## 4. Hierarchical Inheritance

In `Hierarchical Inheritance`, multiple child classes inherit from a single parent class.

📌 Structure

*          Parent
*       /     |    \
*  child1   child2  child3


## Why it is used
* One parent has common functionality.
* Different children use and extend it differently.

## Real-life Example
* Parent: `Employee`
* Child:
     * `Manager`
     * `Developer`
     * `Tester`
All are employees but have different roles.

```python
class Employee: # parent
    def work(self):
        print("Employee is working")

class Manager(Employee): # child
    def manage(self):
        print("Manager is managing")

class Developer(Employee):  # child 
    def code(self):
        print("Developer is coding")

class Tester(Employee):  # child 
    def test(self):
        print("Tester is Testing")


m = Manager()
m.work()
m.manage()

d = Developer()
d.work()
d.code()

t = Tester()
t.work()
t.test()
```

## 5. Hybrid Inheritance

`Hybrid Inheritenace` is a combination of two or more inheritance types.
* It may include:
  * Multiple+Multilevel
  * Hierarchical+Multiple

📌 Structure

*     A
*    / \
*   B   C
*    \ /
*     D



Here:
  * B and C inherits from A (Hierarchical)
  * D inherits from Both B and C (Multiple)

### Why it is used
 * When a system requrires complex relaitonships
 * Common in large frameworks and enterprise projects.



```python
class A: # parent
    def showA(self):
        print("Class A")
class B(A): # child
    def showB(self):
        print("Class B")
class C(A): # child
    def showC(self): 
        print("Class C")
class D(B,C):  # B, C is parent and D is child
    def showD(self):
        print("Class D")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```
