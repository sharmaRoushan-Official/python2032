### 6. Handling Specific Exceptions

Always prefer handling specific exceptions, not general.

```python
try:
    a = 10
    b = 0
    print(a/b)
except ZeroDivisionError:  # specific error
    print("Cannot divide by zero")
```

### 7. Multiple Except Blocks
You can handle multiple errors

```python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x/y)
except ValueError:  # specific error
    print("Invalid Input! Please enter only numbers.")
except ZeroDivisionError: # specific error
    print("You cannot divide by zero")
```

## 8. Single Except Handling Multiple Exceptions

except(Exception1,Exception2):

```python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x/y)
except(ValueError, ZeroDivisionError):
    print("Invalid Input or Division by Zero")
```

### 9. Using Exception Object (as e)
This is used to print actual error message.

```python
try:
    a = int(input("Enter A:"))
    b = int(input("Enter B: "))
    print(a/b)
except Exception as e:
    print("Error Occurred:",e)
```

### 10. try-except-else Block

else block runs only if no exception occurs.

```python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    result = a/b
except ZeroDivisionError:
    print("Cannot Divided By Zero")
else:
    print("Division Result is:",result)
```

```python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    result = a/b
except Exception as e:
    print(e)
else:
    print("Division Result is:",result)
```

```python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    result = a/b
except:
    print("Error")
else:
    print("Division Result is:",result)
```

Flow:
* if exception -> except runs
* if not exception -> else runs

### 11. try-except-finally Block 

finally block always runs(whether exception happens or not).



```python
try:
    file = open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r')
    content = file.read()
    
except FileNotFoundError:
    print("File not found")
else:
    print(content)
finally:
    print("\nProgram finished execution.")
```

### 12. Real use of finally (Resource Closing)

```python
try:
    file = open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r')
    content = file.read()

except Exception as e:
    print(e)
else:
    print(content)
finally:
    file.close()
    print("File Closed")
```

⚠️ Problem: If file not found, file object will not exit and file.close() will crash

### 13. Nested try-except
when exception handling is needed inside another try:

```python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    try:
        print(a/b)
    except Exception as e:
        print(e)
except:
    print("Outer Block: Invalid input")
```

```python
try:
    a = int(input("Enter A: "))
    b = int(input("Enter B: "))
    try:
        print(a/b)
    except Exception as e:
        print(e)
except Exception as e:
    print(e)
```
