seek()
Move pointer to position.

```python
file = open("C:\\Users\\User\\Desktop\\data.txt",'r')

file.seek(5)
print(file.read())
```

```python
print(file.tell())
print(file.seek(10))
```

```python
print(file.seek(13))
print(file.read())
```

### 15. Hadling File Exception

Sometimes file may not exist.

Use try-except

```python
try:
    file  = open('data2.json','r')
    print(file.read())
except Exception as e:
    print(e)
```

```python
try:
    file  = open('data21.json','r')
    print(file.read())
except:
    print("File Not Found")
```

### Search Word in the file



```python
word = "Amit"

with open("students.txt",'r') as file:
    content = file.read()
    for name in file:
        name1 = name.strip()
        if name1:
            print(name1)

if word.capitalize() in content:
    print("Word Found:",word)
else:
    print("Word Not found:",word)
```
