# 7. Conditions in Python


### 1. What is a condition Statement?

 * A `conditional statement` allows a program to make decidions based on conditions(True/False).

 * In Python, we use:

   * if
   * elif
   * else

### 2. Basic If Statement

✔️ Syntax

```python
age = 30    # variable which contains 18 value

if age>=18:# condition [if condition is true/False]
    print("You are elegible to vote")  # statement  [executed/Not executed]
    # (Indentation)
    
```

Key Points:

* Condition Must return True/False
* Indentation is mandatory
* No parentheses required(unlike c/Java)

```python
age>=18
```

## 3. if-else statement


✔️ syntax

```python
age = 15

if age>=18:
    print("Adult")
else:
    print("Minor")
```

Key Points

* else executes when condition is false
* Only one else allowed per if 

### 4. ladder if-elif-else Ladder

✔️ syntax

if condition 1:
    
    # block 1
    
elif condition 2:
    
    # block 2
    
elif condition 3:
    
    # block 3
    
else:
    
    # default block

```python
marks = 91  # A,B,C,D

if marks>=90:
    print("Grade A")
elif marks>=70:
    print("Grade B")
elif marks>=50:
    print("Grade C")
elif marks>=40:
    print("Grde D")
else:
    print("Fail")
```

### Key points
* Multiple conditions can be checked
* Only first True condition executes
* else is optional.

### 5. Nested if statements



```python
age = 20

citizen = not(not(not(False)))

name = "rohit"
if age>=18:
    if citizen:
        print("Elegible to vote")
        if name == "Rohit":
            print("Myself Rohit")
        else:
            print("Ankit")
    else:
        print("Not a Citizen")
else:
    print("Under Age")
```

### 6. Ternary Operator


syntax:

```python
age = 21

status = "Adult" if age>18 else "Minor"
print(status)
```

### 12. Using in and not in [membership operator]

```python
fruits = ["apple","Pineapple","Mango","Grapes","Banana"]


if 'Apple'.lower() or "apple".capitalize() in fruits:
    print("Available")
else:
    print("Not Available")
```

### 13. Chained Conditions in One line

```python
x = 14

if 5< x <15:
    print(x,"Valid Range")
else:
    print(x,"Not Valid Range")
```

```python
x = 21

if 5< x <15:
    print(f"{x} Valid Range")  # f-string
else:
    print(f"{x} Not Valid Range")
```

### 14. Multiple Conditions in One Line
