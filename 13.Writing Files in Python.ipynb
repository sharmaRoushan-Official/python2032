### 13. Writing Files in Python

#### 1. What is File Handling in Python

`File Handling` means working with files stored on the computer using Python.

With Python we Can:
* Create a file
* Read data from a file
* Write data to a file
* Append data to a file
* Delete a file

files are stored on secondary storage(hard disk) and help us save data permanently.

Example Uses:
* Saving Our data
* Storing logs
* Writing reports
* Saving configuration settings.


### 2. Basic Steps for Writing a File

When writing to a file, Python follows `three main steps:`

* 1️⃣ Open the file
* 2️⃣ Write data into the file
* 3️⃣ Close the file

```python
file = open("C:\\Users\\User\\Desktop\\data.txt",'r')
content = file.read()
print(content)
```

```python
file = open("C:\\Users\\User\\Desktop\\data.txt",'w')
file.write("Hello World\nCongratulations")
file.close()
```

## Explanation

* open()- Opens the file
* "C:\\Users\\User\\Desktop\\data.txt"- path with fileName
* 'w'-> write mode
* write() -> writes data in the file
* close() -> closes the file

```python
file = open("C:\\Users\\User\\Desktop\\data.txt",'r')
content = file.read()
print(content)
```

### File Opening Modes 
---
|Mode|Meaning|Description|
|---|---|---|
|"r"|Read|Opens file for reading|
|"w"|Write|Creates file or overwrites exisiting file|
|"a"|Append|Adds data at tend of the file|
|'x'|Create|Creates new file, error if exists|
|"b"|Binary|Used for images/files|
|"t"|Text|Default Mode|

```python
open("file.txt","r") # read
open("file.txt","w")  # write
open("file.txt",'a') # append
open("file.txt","b") # binary
open('file.txt','t') # default mode
```

### Writing Data to a File

Method 1: write()

write(): writes one string to a file

```python
file = open("C:\\Users\\User\\Desktop\\data.txt","w")
file.write("Name: Ram\n")
file.write("Age:22\n")
file.write("Course:Python\n")
file.close()
```

### Writing Multiple Lines

writelines():Writes Multiple lines at once.

```python
file = open("C:\\Users\\User\\Desktop\\data.txt","w")

lines = [
    "Python Programming\n",
    "File handling\n",
    "Object Oriented Programming"
]
file.writelines(lines)
file.close()
```

## Using "with" Statement

Instead of manully closing files, Python Provides with statement.

Benefits:

* Automatically closes file
* Safer
* Cleaner code

```python
with open("C:\\Users\\User\\Desktop\\data.txt","w") as file:
    file.write("Welcome to Python File Handling.")
```

After the block ends- file automatically closes.


### Append Data to File

Append Means `add new data without deleting old data.`

Mode used: "a"

Example

```python
with open("C:\\Users\\User\\Desktop\\data.txt","a") as file:
    file.write("\nNew user logged in\n")
```

Example of Data Writer

```python
name = input("Enter Student Name: ")
age = input("Enter age: ")
course = input("Enter course: ")

with open("C:\\Users\\User\\Desktop\\data.txt","a") as file:
    file.write("\nName: "+name+"\n")
    file.write("Age: "+age+"\n")
    file.write("Course: "+course+"\n")
    file.write(str(100))
    # file.write(100)
    file.write("-------------------\n")
print("Data Saved Successfully")
```
