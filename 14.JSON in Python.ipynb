## 2. JSON vs Python Dictionary

JSON looks like a Python dictionary but there are differences.

|Feature|JSON|Python Dictionary|
|---|---|----|
|Format|Text format|Python Object|
|Quotes|Key must be in double quotes ""|Key can be in single or double quotes|
|Data Type|Limited Data type|Supports many types|
|Used|Data transfer/storage|Python Logic|

python dictionary:



```python
data = {
    'name':"Arun",
    "age":25,
    'email':"arun@gmail.com"
}
data
```

```python
# logic of json key must be in " " quote
{"name":"Arun","age":25}
```

## 5. Convert Python Dictionary to JSON(Serialization)

This is called serialization.

json.dumps() -> Convert Python Object to JSON string

```python
data = {
    "name":"Amit",
    'age':30,
    'city':"Hydrabad"
}
print(data,type(data))
```

```python
json_data = json.dumps(data) # here is convert python dictionary to json
print(json_data)
print(type(json_data))
```

## 6. Convert JSON to Python Dictionary [Deseriaization]

This is called Deserialization.

json.loads() -> Convert JSON string to Python Object

```python
json_string = '{"name":"Amit","age":30,"city":"Hydrabad"}'
print(json_string,type(json_string))
```

```python
python_dict = json.loads(json_string)
print(python_dict)
print(type(python_dict))
```

```python
data = {
    "name":"Amit",
    'age':30,
    'city':"Hydrabad"
}
print(data,type(data))
```

```python
data['name'] = "Amit Singh"
data
```

```python
json_str = json.dumps(data)
json_str
```

```python
python_Obj = json.loads(json_str)
python_Obj
```

```python
python_Obj['age'] = 80
python_Obj
```

```python
new_json = json.dumps(python_Obj)
new_json
```

## 7. Writing JSON DATA to a File

json.dump() :- Write Python object to a JSON file

```python
data = {
    "id":101,
    'name':"Priya",
    "skills":["Python","Django","SQL"]
}
print(data)
```

```python
with open("C:\\Users\\User\\Desktop\\data.json",'w') as file:
    json.dump(data,file)
```

### Reading JSON data from a file 

json.load() -> Read JSON file into Python Object



```python
with open("C:\\Users\\User\\Desktop\\data.json",'r') as file:
    data  = json.load(file)
print(data)
print(type(data))
```

### 9. dumps() vs dump() and loads() vs load()

This is very important points

---

|__Function__|__Used For__|__Input__|__Output__
|----|----|----|----|
|json.dumps()|Python->JSON string|Python Object|JSON String|
|json.dump()|Python->JSON file|Python Object+File|Writes to file|
|json.loads()|Json string->Python|JSON string| Python Object|
|json.load()|JSON file-> Python| file|Python Object
